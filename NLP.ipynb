```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

text = "Natural Language Processing (NLP) is amazing. It helps computers understand text."
print(sent_tokenize(text))
# ['Natural Language Processing (NLP) is amazing.', 'It helps computers understand text.']

words = word_tokenize(text)
print(words)
# ['Natural', 'Language', 'Processing', '(', 'NLP', ')', 'is', 'amazing', '.', 'It', 'helps', 'computers', 'understand', 'text', '.']
```
